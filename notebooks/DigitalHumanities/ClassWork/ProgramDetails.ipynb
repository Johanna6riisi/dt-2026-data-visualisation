Create a chart showing who puts in the most effort

```python
import pandas as pd 
df = pd.read_csv("Competencies.csv")
df.head() # Show the first few rows
df.describe() #Show some basic statistics
```

```python
import matplotlib.pyplot as plt
avg_hours= df.groupby("Student")["Hours_Studied"].mean()
avg_hours.plot(kind="bar", x="Student", y="Hours_Studied", color="#FFC0CB")
#colors= plt.cm.plasma(range(len(avg_hours)))
plt.title("Bar chart for the most effort put into studying")
plt.ylabel("Hours studied")
plt.show()
```

```python
plt.scatter(df["Hours_Studied"], df["Grade"], s=df["Hours_Studied"]*30, alpha=0.6, color="#B67ED6")
plt.title("Grade vs hours studied")
plt.xlabel("Hours studied")
plt.ylabel("Grade")
```

```python
plt.Figure(figsize=(8,6))
df[["Hours_Studied", "Grade"]].boxplot()
plt.title("Distribution of study hours and grades")
plt.show
```
