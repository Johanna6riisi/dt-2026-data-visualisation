```python
import pandas as pd 
df = pd.read_csv("History.csv")
df.head()
df.describe()
```

```python
import matplotlib.pyplot as plt
incident_type=df["Type"].value_counts()
incident_type.plot(kind="bar", color="#FFC0CB")
#colors= plt.cm.plasma(range(len(avg_hours)))
plt.title("Number of incidents by type")
plt.ylabel("Number of incidents")
plt.xlabel("Type")
plt.show()
```

The bar chart shows us that between the years 1215-1807 there were more political and military type incidents in the UK.

```python
affected_counts = df["Affected"].value_counts()
affected_counts.plot(kind="pie", autopct="%1.1f%%")
plt.title("Incidents by affected group")
plt.ylabel("")
plt.show()
```

This pie chart shows us what kinds of societal groups in the UK were affected during the years 1215 to 1807 by important historical incidents. This chart shows us that the English population was the most affected.

```python
year_counts = df["Year"].value_counts().sort_index()
year_counts.plot(kind="line", marker="o")
plt.title("Number of incidents by year")
plt.xlabel("Year")
plt.ylabel("Number of incidents")
plt.show()
```

This line chart shows us that there were more important historical incidents between the 15th and 17th century.
